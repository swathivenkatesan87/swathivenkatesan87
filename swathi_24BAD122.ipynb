```python
print("swathi-24BAD122")
# WORLD HAPPINESS 2017 – FULL ANALYSIS

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/happiness_2017.csv")

cols = df.columns.str.lower()
score_col = df.columns[cols.str.contains("score")][0]
rank_col = df.columns[cols.str.contains("rank")][0]
country_col = df.columns[cols.str.contains("country")][0]
gdp_col = df.columns[cols.str.contains("gdp")][0]

print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())

df.hist(figsize=(12,10))
plt.show()

plt.figure(figsize=(12,6))
sns.boxplot(data=df.select_dtypes(include="number"))
plt.xticks(rotation=90)
plt.show()

top10 = df.sort_values(by=score_col, ascending=False).head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=score_col, y=country_col, data=top10)
plt.show()

plt.figure(figsize=(10,6))
plt.plot(df[rank_col], df[score_col])
plt.xlabel(rank_col)
plt.ylabel(score_col)
plt.show()

plt.figure(figsize=(8,6))
sns.scatterplot(x=gdp_col, y=score_col, data=df)
plt.show()

plt.figure(figsize=(10,8))
sns.heatmap(df.select_dtypes(include="number").corr(), annot=True, cmap="coolwarm")
plt.show()

sns.pairplot(df.select_dtypes(include="number"))
plt.show()
```
